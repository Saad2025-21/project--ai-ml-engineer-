```python
import matplotlib.pyplot as plt 
```

```python
# Temperature over cities
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]

fig,ax = plt.subplots(2,2,figsize=(7,5))
count =0
for i in range(0,2):
    for j in range(0,2):
        ax[i,j].plot(days,temperatures[count])
        ax[i,j].set_title(cities[count])
        
        count+=1
fig.suptitle("Temperature over cities")       
fig.supylabel("Temperatures")
fig.supxlabel("Days")
fig.tight_layout()
```

```python
fig,ax = plt.subplots(2,2)
count =0
for i in range(0,2):
    for j in range(0,2):
        ax[i,j].bar(days,temperatures[count])
        ax[i,j].set_title(cities[count])
        count+=1
        
fig.suptitle("Temperature over cities")       
fig.supylabel("Temperatures")
fig.supxlabel("Days")
fig.tight_layout()
```
